```python
import spacy
from nltk import ngrams
nlp=spacy.load('en_core_web_sm')
doc=nlp("ai is simulation of humans intelligence by machine it includes expects system nlp")
bigrams=ngrams(doc,2)
print(list(bigrams))
```
